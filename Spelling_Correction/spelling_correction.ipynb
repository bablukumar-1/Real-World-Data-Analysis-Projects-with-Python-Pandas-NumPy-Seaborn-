## Spelling Correction Project


```python
dict =["recive","believe","friend","learning","machine","python","science","data","school","college","analysis"]

dict
```

```python
# Create character n-grams
def char_ngrams(word, n=2):
    word = word.lower()
    return [word[i:i+n] for i in range(len(word)-n+1)]
```

```python
# Jaccard similarity
def jaccard_similarity(a, b):
    set1 = set(a)
    set2 = set(b)
    return len(set1 & set2) / len(set1 | set2)
```

```python
# Function to calculate spelling score
def correct_spelling(word, dictionary, n=2):
    word_ngrams = char_ngrams(word, n)
    best_match = None
    best_score = 0
    
    for correct_word in dictionary:
        correct_ngrams = char_ngrams(correct_word, n)
        score = jaccard_similarity(word_ngrams, correct_ngrams)
        
        if score > best_score:
            best_score = score
            best_match = correct_word
            
    return best_match, best_score
```

```python
# wrong_words = ["recivens","fryend","learning","machene","scinse"]
test_words = [" ", "a", "zzzzzz", "12345", "Machine", "SCIENSE"]

for word in test_words:
    match, score = correct_spelling(word, dict)
    print(f"{word} ----> {match}  (score={score:.2f})")
```
